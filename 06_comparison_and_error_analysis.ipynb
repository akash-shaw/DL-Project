* McNemar significance tests and bootstrap confidence intervals,
* tuning / ablation summary, efficiency comparison,
* error analysis: by review length, by linguistic phenomenon, error overlap and qualitative failure cases,
* all figures used in the report (`results/figures/`).

```python
# ---- Setup: runs locally (from notebooks/) or on Google Colab ----
import sys
from pathlib import Path

IN_COLAB = "google.colab" in sys.modules
if IN_COLAB:
    # Keep data and results on Drive so every notebook in the series sees them.
    from google.colab import drive
    drive.mount("/content/drive")
    ROOT = Path("/content/drive/MyDrive/dl-sentiment-project")
else:
    ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

RAW = ROOT / "data" / "raw"
PROC = ROOT / "data" / "processed"
RESULTS = ROOT / "results"
FIG = RESULTS / "figures"
for d in (RAW, PROC, RESULTS, FIG):
    d.mkdir(parents=True, exist_ok=True)
print("project root:", ROOT)
```

```python
import json, re
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
import pandas as pd
from scipy.stats import binomtest, chi2

# fixed colour per model (never re-assigned by rank)
MODELS = {"tfidf_logreg": ("TF-IDF + LR", "#2a78d6"), "text_cnn": ("Text CNN", "#eb6834"),
          "lstm": ("BiLSTM", "#1baf7a"), "distilbert": ("DistilBERT", "#eda100")}
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#e4e3df"
plt.rcParams.update({"font.family": "serif", "font.serif": ["Times New Roman", "DejaVu Serif"], "font.size": 10,
                     "axes.spines.top": False, "axes.spines.right": False, "axes.edgecolor": MUTED,
                     "xtick.color": MUTED, "ytick.color": MUTED})

res = {k: json.loads((RESULTS / k / "metrics.json").read_text()) for k in MODELS if (RESULTS / k / "metrics.json").exists()}
probs = {k: np.load(RESULTS / k / "test_probs.npy") for k in res}            # primary seed (42)
test = pd.read_csv(PROC / "test.csv")
y = test["label"].values
pred = {k: (p >= .5).astype(int) for k, p in probs.items()}
correct = {k: pred[k] == y for k in res}
name = lambda k: MODELS[k][0]
print("models:", [name(k) for k in res])
```

## 1. Main comparison table (common protocol, 25,000 test reviews)

```python
rows = []
for k, r in res.items():
    m, s = r["test_mean"], r["test_std"]
    n = len(r["runs"])
    fmt = lambda key: f"{m[key]:.4f} Â± {s[key]:.4f}" if n > 1 else f"{m[key]:.4f}"
    rows.append({"Model": name(k), "Family": r["family"], "Seeds": n,
                 "Accuracy": fmt("accuracy"), "Precision": fmt("precision"), "Recall": fmt("recall"),
                 "F1": fmt("f1"), "ROC-AUC": fmt("roc_auc"),
                 "Trainable params": r["trainable_params"],
                 "Train time / run (s)": np.mean([x["train_time_s"] for x in r["runs"]]),
                 "Test inference (s)": np.mean([x["infer_time_s"] for x in r["runs"]]),
                 "_acc": m["accuracy"], "_f1": m["f1"], "_f1_std": s["f1"], "_auc": m["roc_auc"]})
comp = pd.DataFrame(rows)
comp["Reviews / s (inference)"] = (len(y) / comp["Test inference (s)"]).round(0)
comp.to_csv(RESULTS / "comparison.csv", index=False)
display(comp.drop(columns=[c for c in comp if c.startswith("_")]).style.format(
    {"Train time / run (s)": "{:.0f}", "Test inference (s)": "{:.1f}", "Trainable params": "{:,}"}))
```

## 2. Statistical significance
McNemar's test (continuity-corrected Ï‡Â², exact binomial when discordant pairs < 25) on the primary-seed predictions, and 95 % bootstrap confidence intervals (1,000 resamples) for accuracy.

```python
def mcnemar(a, b):
    n01 = int(np.sum(a & ~b)); n10 = int(np.sum(~a & b))
    if n01 + n10 < 25:
        p = binomtest(n01, n01 + n10, 0.5).pvalue
    else:
        p = chi2.sf((abs(n01 - n10) - 1) ** 2 / (n01 + n10), 1)
    return n01, n10, p

keys = list(res)
sig = []
for i, a in enumerate(keys):
    for b in keys[i + 1:]:
        n01, n10, p = mcnemar(correct[a], correct[b])
        sig.append({"A": name(a), "B": name(b), "only A correct": n01, "only B correct": n10,
                    "acc A - acc B": (n01 - n10) / len(y), "p-value": p, "significant (p<0.05)": p < 0.05})
sig = pd.DataFrame(sig)
sig.to_csv(RESULTS / "mcnemar.csv", index=False)
display(sig.style.format({"acc A - acc B": "{:+.4f}", "p-value": "{:.2e}"}))

rng = np.random.default_rng(0)
idx = rng.integers(0, len(y), (1000, len(y)))
ci = {name(k): np.percentile(correct[k][idx].mean(1), [2.5, 97.5]) for k in keys}
ci = pd.DataFrame(ci, index=["2.5%", "97.5%"]).T
ci.to_csv(RESULTS / "bootstrap_ci.csv")
ci.round(4)
```

## 3. Tuning and ablations (validation-selected; test shown for analysis only)

```python
abl = []
for k, r in res.items():
    for a in r["ablations"]:
        abl.append({"Model": name(k), "Variant": a["variant"], "Val F1": a["val_f1"],
                    "Test acc": a["test_acc"], "Test F1": a["test_f1"],
                    "Final": a["variant"] == r["final_variant"] or a["variant"].startswith(r["final_variant"])})
abl = pd.DataFrame(abl)
abl.to_csv(RESULTS / "ablations.csv", index=False)
display(abl.style.format({"Val F1": "{:.4f}", "Test acc": "{:.4f}", "Test F1": "{:.4f}"}))
for k, r in res.items():
    print(f"{name(k):12s} final: {r['final_variant']}")
```

## 4. Figures for the report

```python
# Learning curves (primary seed) of the neural models
neural = [k for k in res if res[k]["runs"][0]["history"]]
fig, axes = plt.subplots(1, 2, figsize=(8, 2.9))
for k in neural:
    h = pd.DataFrame(res[k]["runs"][0]["history"]); c = MODELS[k][1]
    axes[0].plot(h.epoch, h.train_loss, color=c, lw=2, label=f"{name(k)} train")
    axes[0].plot(h.epoch, h.val_loss, color=c, lw=2, ls="--")
    axes[1].plot(h.epoch, h.val_acc, color=c, lw=2, marker="o", ms=4, label=name(k))
axes[0].set(xlabel="Epoch", ylabel="BCE loss", title="Loss (solid: train, dashed: val)")
axes[1].set(xlabel="Epoch", ylabel="Validation accuracy", title="Validation accuracy")
axes[1].yaxis.set_major_formatter(mticker.PercentFormatter(1.0, decimals=0))
for a in axes:
    a.grid(axis="y", color=GRID); a.xaxis.set_major_locator(mticker.MaxNLocator(integer=True))
axes[1].legend(frameon=False, fontsize=8)
plt.tight_layout(); fig.savefig(FIG / "training_curves.png", dpi=200); plt.show()

# Confusion matrices
fig, axes = plt.subplots(1, len(res), figsize=(2.4 * len(res), 2.4))
for ax, k in zip(axes, res):
    cm = np.array(res[k]["runs"][0]["test"]["confusion_matrix"])
    ax.imshow(cm, cmap="Blues", vmin=0, vmax=cm.max())
    for i in range(2):
        for j in range(2):
            ax.text(j, i, f"{cm[i, j]:,}", ha="center", va="center", fontsize=9,
                    color="white" if cm[i, j] > cm.max() / 2 else INK)
    ax.set(xticks=[0, 1], yticks=[0, 1], xticklabels=["neg", "pos"], yticklabels=["neg", "pos"], xlabel="Predicted")
    ax.set_title(name(k), fontsize=9)
axes[0].set_ylabel("True")
plt.tight_layout(); fig.savefig(FIG / "confusion_matrices.png", dpi=200); plt.show()

# Accuracy vs. training cost (one point per model)
fig, ax = plt.subplots(figsize=(5.5, 2.9))
for _, r_ in comp.iterrows():
    k = [k for k in res if name(k) == r_["Model"]][0]
    ax.scatter(r_["Train time / run (s)"], r_["_acc"], s=70, color=MODELS[k][1], edgecolor="white", lw=2, zorder=3)
    ax.annotate(r_["Model"], (r_["Train time / run (s)"], r_["_acc"]), xytext=(8, -3), textcoords="offset points", fontsize=9)
ax.set_xscale("log"); ax.set(xlabel="Training time per run (s, log scale)", ylabel="Test accuracy")
ax.yaxis.set_major_formatter(mticker.PercentFormatter(1.0, decimals=0)); ax.grid(color=GRID)
plt.tight_layout(); fig.savefig(FIG / "accuracy_vs_cost.png", dpi=200); plt.show()
```

## 5. Error analysis
### 5.1 Accuracy by review length

```python
n_words = test["text"].str.split().str.len().values
labels = ["<=128", "129-256", "257-512", ">512"]
bucket = pd.cut(n_words, [0, 128, 256, 512, np.inf], labels=labels)
by_len = pd.DataFrame({name(k): pd.Series(correct[k]).groupby(bucket, observed=False).mean() for k in res})
by_len.insert(0, "reviews", pd.Series(bucket).value_counts().reindex(labels).values)
by_len.to_csv(RESULTS / "accuracy_by_length.csv")
display(by_len.round(4))

fig, ax = plt.subplots(figsize=(6.5, 2.9))
x = np.arange(len(labels))
for k in res:
    ax.plot(x, by_len[name(k)], color=MODELS[k][1], lw=2, marker="o", ms=6, label=name(k))
    ax.annotate(f"{by_len[name(k)].iloc[-1]:.1%}", (x[-1], by_len[name(k)].iloc[-1]), xytext=(8, 0),
                textcoords="offset points", va="center", fontsize=8)
ax.set(xticks=x, xticklabels=[f"{l}\n(n={n:,})" for l, n in zip(labels, by_len.reviews)], xlim=(-.3, len(labels) - .35),
       ylabel="Test accuracy", xlabel="Review length (words, after cleaning)")
ax.yaxis.set_major_formatter(mticker.PercentFormatter(1.0, decimals=1)); ax.grid(axis="y", color=GRID)
ax.legend(frameon=False, fontsize=8, loc="lower left")
plt.tight_layout(); fig.savefig(FIG / "accuracy_by_length.png", dpi=200); plt.show()
```

### 5.2 Accuracy on linguistic phenomena
Subsets of the test set identified with simple patterns: contrast/concession markers, negation, explicit numeric ratings, and sarcasm-style cues.

```python
text = test["text"]
subsets = {
    "all reviews": np.ones(len(y), bool),
    "contrast (but/however/although/though/yet)": text.str.contains(r"\b(?:but|however|although|though|yet)\b").values,
    "no contrast marker": ~text.str.contains(r"\b(?:but|however|although|though|yet)\b").values,
    "negation-dense (>= 5 negations)": (text.str.count(r"\b(?:not|no|never|nothing|n't)\b|n't\b") >= 5).values,
    "explicit rating (x/10, x out of 10)": text.str.contains(r"\b\d+(?:\.\d)?\s*(?:/|out of)\s*10\b").values,
    "short (<= 128 words)": n_words <= 128,
    "long (> 256 words)": n_words > 256,
}
phen = pd.DataFrame({name(k): {s: correct[k][m].mean() for s, m in subsets.items()} for k in res})
phen.insert(0, "reviews", [int(m.sum()) for m in subsets.values()])
phen.to_csv(RESULTS / "accuracy_by_phenomenon.csv")
phen.round(4)
```

### 5.3 Error overlap and bias

```python
keys = list(res)
stack = np.stack([correct[k] for k in keys])
overlap = {
    "errors per model": {name(k): int((~correct[k]).sum()) for k in keys},
    "false positives / false negatives": {name(k): {"FP": int(((pred[k] == 1) & (y == 0)).sum()),
                                                    "FN": int(((pred[k] == 0) & (y == 1)).sum())} for k in keys},
    "misclassified by all models": int((~stack.any(0)).sum()),
    "correct by all models": int(stack.all(0).sum()),
    "pairwise agreement on correctness": {f"{name(a)} vs {name(b)}": float(np.mean(correct[a] == correct[b]))
                                          for i, a in enumerate(keys) for b in keys[i + 1:]},
    "oracle accuracy (any model correct)": float(stack.any(0).mean()),
    "ensemble accuracy (mean probability of all models)":
        float(((np.stack([probs[k] for k in keys]).mean(0) >= .5).astype(int) == y).mean()),
}
print(json.dumps(overlap, indent=2))
```

### 5.4 Qualitative failure cases

```python
def examples(mask, k_ref=None, n=6):
    idx = np.where(mask)[0]
    conf = np.abs(np.mean([probs[k] for k in keys], 0) - .5) if k_ref is None else np.abs(probs[k_ref] - .5)
    idx = idx[np.argsort(-conf[idx])][:n]
    return [{"true": "positive" if y[i] else "negative",
             **{f"p({name(k)})": round(float(probs[k][i]), 3) for k in keys},
             "words": int(n_words[i]), "text": test["text"].iloc[i][:450]} for i in idx]

cases = {"all models wrong (most confident)": examples(~stack.any(0))}
if "distilbert" in res:
    others = [k for k in keys if k != "distilbert"]
    cases["only DistilBERT correct"] = examples(correct["distilbert"] & ~np.stack([correct[k] for k in others]).any(0), "distilbert")
    cases["DistilBERT wrong, TF-IDF correct"] = examples(~correct["distilbert"] & correct["tfidf_logreg"], "distilbert")
for title, ex in cases.items():
    print("=" * 100, "\n", title)
    for e in ex[:4]:
        print(f"- true={e['true']} words={e['words']} " + " ".join(f"{k}={v}" for k, v in e.items() if k.startswith("p(")))
        print("  ", e["text"][:300], "...\n")

(RESULTS / "error_analysis.json").write_text(json.dumps({**overlap, "examples": cases}, indent=2))
```
